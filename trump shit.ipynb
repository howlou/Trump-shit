```python
import json
from pathlib import Path

import pandas as pd

data_dir = Path.cwd() / 'data'
file_path = data_dir / 'Trump_Labeled_Combined_Rev_Speeches_Final_9-24-2024.json'

if not file_path.is_file():
    raise FileNotFoundError(
        f'Missing dataset: {file_path}. See the README for setup instructions.'
    )

with file_path.open('r', encoding='utf-8') as file:
    data = json.load(file)
```

```python
df = pd.DataFrame(data)
df.to_csv(data_dir / 'Trump_Labeled_Combined_Rev_Speeches_Final_9-24-2024.csv', index=False, encoding='utf-8')
```
